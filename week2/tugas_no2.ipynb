### Analisis Previous Purchases

Analisis statistik deskriptif dilakukan pada kolom **Previous Purchases**
menggunakan fungsi `describe()`.

Hasil analisis meliputi:

- Mean: nilai rata-rata jumlah pembelian sebelumnya.
- Std: standar deviasi yang menunjukkan tingkat penyebaran data dari nilai rata-rata.
- Min jumlah pembelian sebelumnya yang paling rendah.
- Q1 (25%) nilai kuartil pertama, yaitu batas 25% data terendah.
- Q2 (50%) nilai median atau nilai tengah data.
- Q3 (75%) nilai kuartil ketiga, yaitu batas 75% data.
- Max jumlah pembelian sebelumnya yang paling tinggi.

Analisis ini digunakan untuk mengetahui karakteristik dan persebaran
jumlah pembelian sebelumnya pada pelanggan.

```python
previous = df['Previous Purchases']

print("Mean :", previous.mean())
print("Std  :", previous.std())
print("Min  :", previous.min())
print("Q1   :", previous.quantile(0.25))
print("Q2   :", previous.quantile(0.50))
print("Q3   :", previous.quantile(0.75))
print("Max  :", previous.max())
```

### Hasil Statistik Previous Purchases

Nilai statistik deskriptif pada kolom **Previous Purchases** digunakan
untuk mengetahui nilai rata-rata, penyebaran, nilai minimum, kuartil,
median, dan nilai maksimum dari jumlah pembelian sebelumnya.

Nilai **mean** menunjukkan rata-rata pembelian sebelumnya, sedangkan
**standard deviation (std)** menunjukkan seberapa besar variasi data
terhadap nilai rata-ratanya.

Q1, Q2, dan Q3 digunakan untuk mengetahui posisi distribusi data,
sedangkan min dan max menunjukkan rentang nilai terendah hingga tertinggi.

```python
df[['Age', 'Purchase Amount (USD)']].corr()
```

### Korelasi Age vs Purchase Amount (USD)

Analisis korelasi dilakukan untuk mengetahui hubungan antara **Age**
dan **Purchase Amount (USD)**.

Fungsi `corr()` digunakan untuk menghitung koefisien korelasi antara
kedua variabel.

Nilai korelasi berada pada rentang **-1 hingga 1**:

- Nilai mendekati **1** menunjukkan hubungan positif yang kuat.
- Nilai mendekati **-1** menunjukkan hubungan negatif yang kuat.
- Nilai mendekati **0** menunjukkan hubungan yang sangat lemah atau hampir tidak ada.

Berdasarkan nilai korelasi yang diperoleh dari dataset, hubungan antara
usia pelanggan dan jumlah pembelian dapat dianalisis berdasarkan
arah dan kekuatan nilai korelasinya.

Analisis korelasi selanjutnya dilakukan untuk mengetahui hubungan antara

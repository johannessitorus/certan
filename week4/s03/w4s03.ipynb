```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris
iris = load_iris()
```

## Memuat Dataset Iris

Dataset Iris diambil langsung dari Scikit-Learn, kemudian dikonversi menjadi DataFrame Pandas.

```python
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df["Species"] = iris.target_names[iris.target]

X = df.drop(columns=["Species"])
y = df["Species"]

print (X)
```

```python
y
```

## Encoding Label

Label species masih berupa teks. Label tersebut diubah menjadi angka menggunakan LabelEncoder dari Scikit-Learn agar dapat digunakan oleh model.

```python
from sklearn.preprocessing import LabelEncoder
lab = LabelEncoder()
y = lab.fit_transform(y)
y
```

## Membagi Data Training dan Testing

Pembagian data mengikuti langkah pada modul, yaitu 80% data training dan 20% data testing dengan random_state=0.

```python
from sklearn.model_selection import train_test_split

X_train, X_test = train_test_split(X, test_size=0.2, random_state=0)
y_train, y_test = train_test_split(y, test_size=0.2, random_state=0)
```

```python
from sklearn.naive_bayes import GaussianNB

model = GaussianNB()
NB_model = model.fit(X_train, y_train)
```

```python
result = NB_model.predict(X_test)
result
```

```python
from sklearn.metrics import accuracy_score
print('Accuracy of data testing: ', accuracy_score(y_test, result)*100)
```

## Prediksi Data Baru pada Modul

Modul memberikan contoh input `[5.1, 3.5, 1.4, 0.2]`. Karena model dilatih dengan DataFrame yang memiliki nama kolom, input baru juga dibuat menjadi DataFrame dengan nama kolom yang sama.

```python
input_data = [[5.1, 3.5, 1.4, 0.2]]
result = NB_model.predict(input_data)
prediction_label = lab.inverse_transform(result)
print('Hasil prediksi untuk input baru adalah: ', prediction_label)
```

```python
input_data = pd.DataFrame(np.array([[5.1, 3.5, 1.4, 0.2]]), columns=iris.feature_names)
result = NB_model.predict(input_data)
prediction_label = lab.inverse_transform(result)
print('Hasil prediksi untuk input baru adalah:', prediction_label)
```

# BAGIAN B — TUGAS

## Prediksi Data Baru

Tugas pada modul adalah memprediksi species dari data baru berikut:

[7.3, 3.8, 4.6, 2.3]


Model yang digunakan tetap GaussianNB yang telah dilatih pada bagian praktikum.

```python
import numpy as np
import pandas as pd

try:
    iris
    lab
    NB_model
except NameError:
    from sklearn.datasets import load_iris
    from sklearn.preprocessing import LabelEncoder
    from sklearn.model_selection import train_test_split
    from sklearn.naive_bayes import GaussianNB

    iris = load_iris()
    X = pd.DataFrame(iris.data, columns=iris.feature_names)
    lab = LabelEncoder()
    y = lab.fit_transform(iris.target_names[iris.target])

    X_train, X_test = train_test_split(X, test_size=0.2, random_state=0)
    y_train, y_test = train_test_split(y, test_size=0.2, random_state=0)

    NB_model = GaussianNB()
    NB_model.fit(X_train, y_train)

input_tugas = pd.DataFrame(
    np.array([[7.3, 3.8, 4.6, 2.3]]),
    columns=iris.feature_names
)

hasil_tugas = NB_model.predict(input_tugas)
prediksi_tugas = lab.inverse_transform(hasil_tugas)

hasil_tabel = input_tugas.copy()
hasil_tabel['Prediksi Species'] = prediksi_tugas
display(hasil_tabel)
```

```python
print('Hasil prediksi tugas:', prediksi_tugas[0])

# Probabilitas prediksi untuk melihat keyakinan model
probabilitas = NB_model.predict_proba(input_tugas)[0]
probabilitas_tabel = pd.DataFrame({
    'Species': lab.classes_,
    'Probabilitas': probabilitas
})
display(probabilitas_tabel)
```

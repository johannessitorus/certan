```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# soal nomor 1
print("=== Soal 1: Matriks Random 5x5 ===")
matriks = np.random.randint(1, 100, size=(5, 5))  # elemen random 1-99
print("Matriks:")
print(matriks)

nilai_max = matriks.max()
nilai_min = matriks.min()
nilai_rata2 = matriks.mean()

print("\nNilai maksimum :", nilai_max)
print("Nilai minimum  :", nilai_min)
print("Nilai rata-rata:", nilai_rata2)
```

```python
#soal nomor 2
print("\n=== Soal 2: Grafik 2D ===")

x = np.linspace(-2, 2, 500)
y = np.linspace(-2, 2, 500)
X, Y = np.meshgrid(x, y)

F = (X**6 + 3*X**4*Y**2 - 3*X**4 + 3*X**2*Y**4 - X**2*Y**3
     - 6*X**2*Y**2 + 3*X**2 + Y**6 - 3*Y**4 + 3*Y**2)

plt.figure(figsize=(6, 6))
plt.contour(X, Y, F, levels=[1], colors='crimson')
plt.title('Grafik 2D: Persamaan Implisit')
plt.xlabel('x')
plt.ylabel('y')
plt.axhline(0, color='gray', linewidth=0.5)
plt.axvline(0, color='gray', linewidth=0.5)
plt.grid(True, linestyle='--', alpha=0.5)
plt.gca().set_aspect('equal')
plt.savefig('soal2_grafik2d.png', dpi=150)
plt.show()

print("Grafik 2D disimpan sebagai soal2_grafik2d.png")

```

```python
print("\n=== Soal 3: Grafik 3D ===")
 
from mpl_toolkits import mplot3d  # noqa: F401
 
x = np.linspace(-2, 2, 200)
y = np.linspace(-2, 2, 200)
X, Y = np.meshgrid(x, y)
Z = (5 * Y / 4 - np.abs(X)) ** 2 + X ** 2
 
fig = plt.figure(figsize=(7, 6))
ax = plt.axes(projection='3d')
ax.plot_surface(X, Y, Z, cmap='viridis', edgecolor='none')
ax.set_title('Grafik 3D: z = (5y/4 - |x|)^2 + x^2')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('z')
plt.savefig('soal3_grafik3d.png', dpi=150)
plt.show()
 
print("Grafik 3D disimpan sebagai soal3_grafik3d.png")
 

```
